```python
def check_sum_to_12(elements, element_to_states):

    states_lists = [element_to_states[el] for el in elements]


    for combo in product(*states_lists):
        if sum(combo) == 12:
            return True, combo  
    return False, None
```

```python
element_to_states = {}
for state, elements in oxidation_states.items():
    for el in elements:
        element_to_states.setdefault(el, []).append(state)
```

```python
from itertools import product

rows = []

for a1, a2, b1, b2 in product(a1_list, a2_list, b1_list, b2_list):
    formula = None
    type_compound = None

    if a1 == a2 and b1 == b2 and a1 != b1:
        formula = f"{a1}{b1}O3"
        type_compound = 'ABO3'
    elif len({a1, a2, b1, b2}) == 4:
        formula = f"{a1}{a2}{b1}{b2}O6"
        type_compound = 'A1A2B1B2O6'
    elif a1 != a2 and b1 == b2 and a1 != b1 and a2 != b1:
        formula = f"{a1}{a2}{b2}O6"
        type_compound = 'A1A2B2O6'
    elif a1 == a2 and b1 != b2 and a1 not in (b1, b2):
        formula = f"{a1}{b1}{b2}O6"
        type_compound = 'A2B1B2O6'

    if formula:
        elements = [a1, a2, b1, b2]
        result, combo = check_sum_to_12(elements, element_to_states)

        if result:
            a1_os, a2_os, b1_os, b2_os = combo
            rows.append((a1, a1_os, b1, b1_os, a2, a2_os, b2, b2_os, formula, type_compound))
        

df = pd.DataFrame(rows, columns=['A1', 'A1_OS', 'B1', 'B1_OS', 'A2', 'A2_OS', 'B2', 'B2_OS', 'Formula', 'Type'])
```
