```python
from sklearn.preprocessing import  OneHotEncoder

onehot_encoder = OneHotEncoder(sparse_output=False)
type_encoded = onehot_encoder.fit_transform(df[['type']])
type_encoded_df = pd.DataFrame(type_encoded, columns=onehot_encoder.get_feature_names_out(['type']))

df_prep = pd.concat([df, type_encoded_df], axis=1)

df_prep = df.drop(columns=['type','A1', 'A2', 'B1', 'B2','Formula'])

display(df_prep.info())
```

```python
col_to_drop_for_stable = ['X-B2', 'A1_OS', 'P-B2', 'BE-B2','EA-A1','RA-B1', 'EA-A2']
```

```python
for col in col_to_drop_for_stable:
    if col in cols_for_stable:
        cols_for_stable.remove(col)
print(cols_for_stable)
```

```python
temp_df = df_prep[cols_for_stable]
df_prep['stable'] = stable_model.predict(temp_df)
temp_df = df_prep[cols_for_insulator]
df_prep['Insulator'] = insulator_model.predict(temp_df)
temp_df = df_prep[cols_for_bandgap]
df_prep['PBE_bandgap'] = bandgap_model.predict(temp_df)
```

```python
df['stable'] = df_prep['stable']
df['Insulator'] = df_prep['Insulator']
df['PBE_bandgap'] = df_prep['PBE_bandgap']
```
